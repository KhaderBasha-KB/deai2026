# Task6: Create a pyspark dataframe df1 reading the data from the above file using pyspark magic command %pyth

```python
%python

df1 = spark.read \
    .option("header", True) \
    .option("inferSchema", True) \
    .csv("/Volumes/workspace/default/usage_metrics/employee.csv")

display(df1.limit(10))
```

# **Task7:** Write the above dataframe df1 data into a databricks table called 'default.mobile_os_usage' using pyspark magic command %python

```python
%python

df1.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("default.mobile_os_usage")
display(spark.table("default.mobile_os_usage").limit(10))
```

# Task8: Write sql query to display the data loaded into the table 'default.mobile_os_usage' using the pyspark magic command %python

```python
%python

df2 = spark.sql("SELECT * FROM default.mobile_os_usage")

display(df2)
```
