## Task 6 – Create PySpark DataFrame using %python

```python
%python

df1 = spark.read \
    .option("header", True) \
    .option("inferSchema", True) \
    .csv("/Volumes/workspace/default/usage_metrics/employee.csv")

display(df1)
```

## Task 7 – Write DataFrame into Databricks table

```python
%python

df1.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("default.mobile_os_usage")
display(spark.table("default.mobile_os_usage"))
```

## Task 8 – Execute SQL query using %python

```python
%python

df2 = spark.sql("SELECT * FROM default.mobile_os_usage")

display(df2)
```
